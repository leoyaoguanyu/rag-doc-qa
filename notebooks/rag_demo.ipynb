# RAG Demo — Ask Questions Over Documents

End-to-end walkthrough: load docs → chunk → embed → FAISS index →
retrieve → stuff into prompt template. (LLM generation runs in `src/qa.py`.)

```python
import sys
sys.path.insert(0, "..")
from src.ingest import load_documents, chunk_documents
from src.vectorstore import build_vectorstore
from src.prompts import RAG_PROMPT
```

```python
docs = load_documents("../data/sample_docs")
chunks = chunk_documents(docs)
print(f"{len(docs)} documents -> {len(chunks)} chunks")
print("sample chunk:", chunks[0].page_content[:120], "...")
```

```python
store = build_vectorstore(chunks, persist_dir="../vectorstore")
print("index size:", store.index.ntotal)
```

```python
question = "Which vector search library was developed by Meta?"
hits = store.similarity_search(question, k=2)
for h in hits:
    print(f"[{h.metadata['source']}] {h.page_content[:150]}...\n")

context = "\n\n".join(h.page_content for h in hits)
print(RAG_PROMPT.format(context="<retrieved>", question=question))
```
